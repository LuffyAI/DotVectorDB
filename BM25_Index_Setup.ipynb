```python
import os
from dotenv import load_dotenv
from dotvectordb.DotVDB import DotVectorDB

load_dotenv()
db = DotVectorDB("TEST_TAGS31.db", tokenizer_name=os.getenv("TOKENIZER", "gpt-4o-mini"))
await db.open()
try:
    await db.create_BM25_index_for_table("entity")
    await db.create_BM25_index_for_table("relationship")
    await db.create_BM25_index_for_table("chunk")
finally:
    await db.close()
```

```python
from pprint import pprint

await db.open()
try:
    results = await db.search_BM25("chunk", "little fella", top_k=10)
    for result in results:
        print(result)
        print("")
finally:
    await db.close()
```
