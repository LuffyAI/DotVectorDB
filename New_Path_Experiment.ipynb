```python
import os
from dotenv import load_dotenv
from dotvectordb.DotVDB import DotVectorDB

load_dotenv()
db = DotVectorDB("TEST_TAGS31.db", tokenizer_name=os.getenv("TOKENIZER", "gpt-4o-mini"))
await db.open()
await db.graph_retriever.load_graph(verbose=False)

starting_node_ids = ["8334855762169143541"]  

query_graph, relationship_ids = await db.graph_retriever.weight_paths(
    starting_node_ids=starting_node_ids,
    accepted_node_types=[],
    intermediate_goal="Shrek's friends",
    terminal_goal="Shrek's best friend or side kick",
    hop_limit=5,
)
print([(edge["relation_id"], edge["hop"]) for edge in query_graph.es])

await db.close()
```

```python
from dotvectordb.data_structures.path import PathManager

SOURCE_ID = "8334855762169143541"
DESTINATION_ID = "-3301400920416631559"
END_POINT_ID = "353387405037581281"
K = 10

path_manager = PathManager(
    tokenizer=db.tokenizer,
    max_tokens=100000,
    top_k=K,
)

await db.graph_retriever.get_weighted_shortest_paths(
    graph=query_graph,
    source=SOURCE_ID,
    destination=DESTINATION_ID,
    end_point=END_POINT_ID,
    PathManager=path_manager,
    k=K,
)

[path.path_text for path in path_manager.path_buffer.values()]
```
