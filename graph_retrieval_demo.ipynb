```python
from dotvectordb.plugins.llm_providers.openai_llms import OpenAiAgent
from dotvectordb.plugins.llm_providers.gemini_llms import GeminiAgent
from dotvectordb.plugins.llm_providers.openrouter_llms import OpenRouterAgent
from dotvectordb.plugins.embedding_providers.openai_embeddings import OpenAIEmbeddingModel
from dotvectordb.DotVDB import DotVectorDB


GPTEmbedding = OpenAIEmbeddingModel()
GPT = OpenAiAgent()
db = DotVectorDB(path="test_db5.db", DIM=1536, embedding_model=GPTEmbedding, language_model=GPT)
await db.open()
await db.graph_retriever.load_graph()
```

```python
# Simple vector search example
output, tokens = await db.vector_search(query="Shrek?",table="entity", top_k=5)
print("Vector Search Output:", output)

# Grab first entity ID from first result
if output:
    first_entity_id = output[0]['entity_id'] 
    print("First Entity ID:", first_entity_id)

# Applying to the graph retrieval algorithm
relationship_ids, edge_ids = await db.graph_retriever.bfs_weight_propagation(starting_node_id=first_entity_id)
print("Relationship vector indexes from BFS Weight Propagation:", relationship_ids)
print("Edge IDs from BFS Weight Propagation:", edge_ids)

await db.graph_retriever.create_temporary_relational_graph(relationship_ids=relationship_ids, verbose=True)
```
